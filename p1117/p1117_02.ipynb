```python
import joblib
from scipy.special import softmax
import numpy as np
model2 = joblib.load('model2.joblib')
```

```python
# 242.0	25.4	30.0	11.52	4.0200
# 290.0	26.3	31.2	12.48	4.3056
# 무게, 길이,대각선, 높이, 넓이를 입력받아
# 분류생선과 확률을 출력하시오.
# 분류 생선 : 93.1%의 확률로 Bream 판별됩니다.

# in_data= input('무게, 길이,대각선, 높이, 넓이를 입력하세요>> 예)242.0,25.4,30.0,11.52,4.0200')
# in_data = '242.0,25.4,30.0,11.52,4.0200'
# in_data = '55,14.7,16.5,6.8475,2.3265'
in_data = '32,13.7,14.7,3.528,1.9992' 
# in_data = '300,35,38.8,5.9364,4.3844' 
in_data = in_data.split(',')
in_data = list(map(float,in_data))
in_data
```

```python
# 예측 - 'Bream', 'Parkki', 'Perch', 'Pike', 'Roach', 'Smelt', 'Whitefish'
result = model2.predict([in_data])
# 확률인데.
result2 = model2.predict_proba([in_data])
if result[0] == 'Bream':
    result3 = result2[0][0]
elif result[0] == 'Parkki':
    result3 = result2[0][1]
elif result[0] == 'Perch':
    result3 = result2[0][2]
elif result[0] == 'Pike':
    result3 = result2[0][3]
elif result[0] == 'Roach':
    result3 = result2[0][4]
elif result[0] == 'Smelt':
    result3 = result2[0][5]
elif result[0] == 'Whitefish':
    result3 = result2[0][6]
print("test : ",result3)

decisions = model2.decision_function([in_data])

print("전체확률 : ")
soft1 = softmax(decisions,axis=1)  #axis =1:행방향 / axis=0:열방향
print(np.round(soft1,decimals=3))

# 분류 생선 : 93.1%의 확률로 Bream 판별됩니다.
print(f"분류 생선 : {result[0]} ( 확률 : {result3*100:.2f}% )")
```
