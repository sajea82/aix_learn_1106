```python
import joblib
import os
from pathlib import Path
import numpy as np
from scipy.special import softmax
```

```python
model_ex = joblib.load('model_ex.joblib')
```

```python
# 290.0	26.3	31.2	12.48	4.3056
in_data = input('데이터 입력 : 290.0,26.3,31.2,12.48,4.3056 :')
in_data = in_data.split(',')
in_data = list(map(float,in_data))
print("입력데이터 : ",in_data)
```

```python
model_ex.classes_
```

```python
result = model_ex.predict([in_data])
result2 = model_ex.predict_proba([in_data])
decisions = model_ex.decision_function([in_data])
soft1 = softmax(decisions,axis=1)
result_all = np.round(soft1,decimals=3)[0]
if result == 'Bream':
    result2 = result_all[0]
elif result == 'Parkki':
    result2 = result_all[1]    
elif result == 'Perch':
    result2 = result_all[2]    
elif result == 'Pike':
    result2 = result_all[3]    
elif result == 'Roach':
    result2 = result_all[4]    
elif result == 'Smelt':
    result2 = result_all[5]    
elif result == 'Whitefish':
    result2 = result_all[6]    

print(f'예측 생선 : {result[0]} (예측확률 : {result2}) ')
print('전체확률 : ',result_all)
```
