```python
df = pd.read_csv('data/perch_full.csv')
df.head(2)
```

```python
data = df.iloc[:, :-1].values
```

```python
# 데이터 가져오기
# 데이터 전처리
# 데이터 분리
# PolynomialFeatures 다항추가
# degree 추가
# 정규화작업

#######
# 데이터 학습
# 규제 적용후 데이터 학습
# alpha값 변경후 데이터 학습 확인


# 데이터 예측
# 정확도
```

```python
# 학습데이터, 테스트데이터 분리
train_data, test_data, train_label, test_label = train_test_split(df, perch_weight)
```

```python
# 특성값을 추가
poly = PolynomialFeatures(degree=5,include_bias=False)
poly.fit(train_data)
train_poly = poly.transform(train_data)
test_poly = poly.transform(test_data)
```

```python
# 규제 정규화
ss = StandardScaler()
ss.fit(train_poly)
train_scaled = ss.transform(train_poly)
test_scaled = ss.transform(test_poly)
```

```python
alpha_list = [0.001,0.01,0.1,1,10,100]

lasso = Lasso()
lasso.fit(train_scaled,train_label)
score = lasso.score(train_scaled,train_label)
score2 = lasso.score(test_scaled,test_label)
print('train정확도 : ',score)
print('test정확도 : ',score2)
```
