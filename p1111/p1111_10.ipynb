```python
# 그래프 관련 import
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
# matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus']=False

import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neighbors import KNeighborsRegressor
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures  #  다항특성 자동생성
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_absolute_error       # 평균절대값오차 오차범위
from sklearn.model_selection import train_test_split  # 정규화 클래스
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('data/manhattan.csv')
# df = df.drop(columns=['rental_id', 'neighborhood', 'borough'])
df.head(2)
```

```python
data = df.iloc[:,2:-2].values
label = df.iloc[:, 1].values
```

```python
data.shape
```

```python
data
```

```python
train_data, test_data, train_label, test_label = train_test_split(data,label)
```

```python
lr =  LinearRegression()
lr.fit(train_data,train_label)
score = lr.score(train_data,train_label)
score2 = lr.score(test_data,test_label)
print(score)
print(score2)
```

```python
predict = lr.predict(test_data)
print(predict)
score = lr.score(train_data,train_label)
score2 = lr.score(test_data,test_label)
print(score)
print(score2)
```

```python
poly = PolynomialFeatures(degree=2,include_bias=False)
poly.fit(train_data)
train_poly = poly.transform(train_data)
test_poly = poly.transform(test_data)
print(train_poly.shape)
```

```python
lr = LinearRegression()
lr.fit(train_poly,train_label)
score = lr.score(train_poly,train_label)
score2 = lr.score(test_poly,test_label)
print('train정확도 : ',score)
print('test정확도 : ',score2)
```

```python
# 규제 정규화
ss = StandardScaler()
ss.fit(train_poly)
train_scaled = ss.transform(train_poly)
test_scaled = ss.transform(test_poly)
```

```python
# 평균이 0에 가까운지 확인 (각 컬럼별 평균)
print("평균:", np.mean(train_scaled, axis=0))

# 표준편차가 1에 가까운지 확인 (각 컬럼별 표준편차)
print("표준편차:", np.std(train_scaled, axis=0))
```
