```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['axes.unicode_minus']=False
import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso 
from sklearn.metrics import mean_absolute_error   # 평균절대값오차 오차범위
from sklearn.preprocessing import PolynomialFeatures,StandardScaler  
from sklearn.model_selection import train_test_split
```

```python
# 무게 데이터
perch_weight = [5.9, 32.0, 40.0, 51.5, 70.0, 100.0, 78.0, 80.0, 85.0, 85.0, 
110.0, 115.0, 125.0, 130.0, 120.0, 120.0, 130.0, 135.0, 110.0, 
130.0, 150.0, 145.0, 150.0, 170.0, 225.0, 145.0, 188.0, 180.0, 
197.0, 218.0, 300.0, 260.0, 265.0, 250.0, 250.0, 300.0, 320.0, 
514.0, 556.0, 840.0, 685.0, 700.0, 700.0, 690.0, 900.0, 650.0, 
820.0, 850.0, 900.0, 1015.0, 820.0, 1100.0, 1000.0, 1100.0, 
1000.0, 1000.0]
```

```python
# LinearRegression - 예측 : 지속적인 데이터를 가지고 확률을 알아내는 것
# 분류 : 1,0 / 빙어,도미 구분
df = pd.read_csv('data/perch_full.csv')
df.head(2)
```

```python
# 데이터 전처리
# 학습데이터,테스트 데이터 분리
train_data,test_data,train_label,test_label =\
train_test_split(df,perch_weight , random_state=42)
```

```python
# 특성을 추가해서 정확도를 높이는 작업
poly = PolynomialFeatures(degree=17,include_bias=False)
poly.fit(train_data)
train_poly = poly.transform(train_data)
test_poly = poly.transform(test_data)
```

```python
print(train_data.shape)
print(train_poly.shape)
print(poly.get_feature_names_out())
```

```python
# 규제 - 정규화, 규제알고리즘 사용
# 정규화작업 하지 않으면 규제알고리즘을 사용할수 없음.
ss = StandardScaler()
ss.fit(train_poly)
train_scaled = ss.transform(train_poly)
test_scaled = ss.transform(test_poly)
```

```python
# 규제 알고리즘
ridge = Ridge(alpha=0.1)
ridge.fit(train_scaled,train_label)
print("train 정확도 : ",ridge.score(train_scaled,train_label))
print("test 정확도 : ",ridge.score(test_scaled,test_label))
```
