```python
df = pd.read_csv('data/picher_stats_2017.csv')
```

```python
data = df.drop(['선수명', '연봉(2018)'], axis=1)
label = df['연봉(2018)']
```

```python
data = pd.get_dummies(data)
```

```python
train_data, test_data, train_label, test_label = train_test_split(data, label, test_size=0.2, random_state=42)
```

```python
rf = RandomForestClassifier(random_state=42)
scores = cross_validate(rf, train_data, train_label, return_train_score=True, n_jobs=-1)
print("훈련 점수:", np.mean(scores['train_score']))
print("테스트 점수:", np.mean(scores['test_score']))
```
